## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

My lane is ranking/scoring. I will assign a priority score to content items so the content team can decide which pages should be reviewed first. Ranking is suitable because the goal is to prioritize items rather than make a simple yes/no decision.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

I would predict whether a content item is declining based on the observed trend_direction outcome. The label comes from the measured trend in the dataset, not from a rule I define myself.

```python
import pandas as pd

df = pd.read_csv(
    "https://raw.githubusercontent.com/channakeshavareddy-coder/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
)

print(df.columns.tolist())
print(df["trend_direction"].value_counts(dropna=False))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

I will use Precision@K as the success metric. It measures how many of the top-ranked content items actually need attention. A higher Precision@K means the ranking is better at putting useful items first.

```python
print(df["trend_direction"].value_counts())
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

The unit of analysis is one content item. Each row in the dataframe represents one content item identified by content_id, with its performance, content, and trend-related information.

```python
lane_df = df[["content_id", "client_id", "trend_direction", "trend_pct"]]

print(lane_df.head())
print("Rows:", len(lane_df))
print("Columns:", len(lane_df.columns))
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule may miss complex patterns because content performance depends on multiple factors such as traffic, engagement, content age, search demand, and other signals. ML can combine these signals to learn patterns and produce a more useful ranking than a simple if-statement.

```python
print(lane_df[["trend_pct"]].describe())
```
